```python
#reading the contents of a file
with open('sample.txt','r') as file:
    content=file.read()
    print(content)
```

```python
#read line by line from the file
with open('sample.txt','r') as file:
    for line in file:
        print(line)
        
```

```python
#overwriting a file
with open('sample.txt','w') as file:
    file.write('hello World\n')

```

```python
#copying content from source to destination
with open('source.txt','r') as src:
    with open('destination.txt','w') as dest:
        dest.write(src.read())
        
```

```python
#without overwrtting the file
list=['hello!\n','Gayathri here\n','learning file handling in python\n','aiming to be a ai engineer\n']
with open('sample.txt','a') as file:
    file.writelines(list)
```

```python
# counting words in the file
with open('sample.txt','r') as file:
    text=file.read()
    words=text.split()
    print(len(words))
```

```python
#finding and replacing text
with open('sample.txt','r') as file:
    text=file.read()
new_text=text.replace('hello',"hii")
with open('sample.txt','w') as file:
    file.write(new_text)
```

```python
with open('sample.txt','r') as file:
    lines=file.readlines()
    for line in reversed(lines):
        print(line.strip())
```

```python
#counting letters,words,lines
with open('sample.txt','r') as file:
    lines=file.readlines()
    words=sum(len(line.split()) for line in lines)
    characters=sum(len(line) for line in lines)
    print(len(lines))
    print(words)
    print(characters)
```

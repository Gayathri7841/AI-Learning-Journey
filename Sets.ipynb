```python
s1=set(range(1,6))
s2=set(range(2,11,2))
print(s1|s2)
print(s1-s2)
print(s1&s2)
print(s1^s2)
```

```python
s1=set(num**2 for num in range(1,11))
print(s1)
```

```python
even=set()
for i in  s1:
    if i%2==0:
        even.add(i)
print(even)
```

```python
evens={x for x in s1 if x%2==0}
print(evens)
```

```python
## Feature Engineering for date col
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
df['Hour']=df['Arrival_Time'].str.split(':').str[0]
df['Minutes']=df['Arrival_Time'].str.split(':').str[1]
```

```python
df['Hour']=df['Hour'].astype(int)
df['Minutes']=df['Minutes'].astype(int)
```

```python
df.drop('Arrival_Time',inplace=True,axis=1)
```

```python
df.head()
```

```python
df['dept_Hour']=df['Dep_Time'].str.split(':').str[0]
df['dept_Minutes']=df['Dep_Time'].str.split(':').str[1]
df['dept_Hour']=df['dept_Hour'].astype(int)
df['dept_Minutes']=df['dept_Minutes'].astype(int)
```

```python
df.head()
```

```python
df.drop('Dep_Time',inplace=True,axis=1)
```

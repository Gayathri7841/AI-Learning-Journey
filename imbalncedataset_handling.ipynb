```python
normal=df[df['is_fraud']==0]
fraud=df[df['is_fraud']==1]
```

```python
normal.shape
```

```python
fraud.shape
```

```python
from sklearn.utils import resample
```

```python
fraud_upsample=resample(fraud,
                        replace=True,
                        n_samples=len(normal),
random_state=42
)
```

```python
fraud_upsample.shape
```

```python
fraud_downsample=resample(normal,
                         replace=False,
                         n_samples=len(fraud),
                         random_state=42
                         )
```

```python
fraud_downsample.shape
```

```python
oversample_df=pd.concat([normal,fraud_upsample])
```

```python
undersample_df=pd.concat([fraud,fraud_downsample])
```

```python
matrix=[
[1,2,3],
[4,5,6],
[7,8,9]
]
def transpose(m):
    trans=[[0]*3 for i in range(3)]

    for i in range(3):
        for j in range(3):
            trans[i][j]=m[j][i]
    return trans

res=transpose(matrix)
print(res)
print(matrix)
```

```python
matrix=[
[1,2,3],
[4,5,6],
[7,8,9]
]
def flatten(m):
    list=[]
    for i in range(3):
        for j in range(3):
            list.append(m[i][j])
    return list

res=flatten(matrix)
print(res)
print(matrix)
```

```python
n=3
list=[1,2,3,4,5]
def rotate(n,l):
    rotated=l[-n:]+l[:-n]
    return rotated
res=rotate(n,list)
print(res)
```

```python
l1=[1,2,3]
l2=[4,2,5]
def intersect(l1,l2):
    list=[]
    for i in range(len(l1)):
        if(l1[i] in l2 ):
            list.append(l1[i])
    return list
res=intersect(l1,l2)
print(res)
```

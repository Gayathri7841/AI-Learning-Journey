```python
X=df.drop(columns=['selling_price'])
Y=df['selling_price']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=10)
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder(handle_unknown='ignore', sparse_output=False)
```

```python
encoding_list=['seller_type','transmission_type','fuel_type','brand','model']

X_train_encoded=encoder.fit_transform(X_train[encoding_list])
df_encoded=pd.DataFrame(X_train_encoded,columns=encoder.get_feature_names_out(encoding_list),index=X_train.index)
X_train=pd.concat([X_train.drop(columns=encoding_list),df_encoded],axis=1)

X_test_encoded=encoder.transform(X_test[encoding_list])
df_encoded_test=pd.DataFrame(X_test_encoded,columns=encoder.get_feature_names_out(encoding_list),index=X_test.index)
X_test=pd.concat([X_test.drop(columns=encoding_list),df_encoded_test],axis=1)
```

```python
X_train.drop(columns=['car_name'], inplace=True)
X_test.drop(columns=['car_name'], inplace=True)
```
